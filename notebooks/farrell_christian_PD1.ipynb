```python
import time

def timed_call(fn, *args, **kwargs):
    start = time.perf_counter()
    result = fn(*args, **kwargs)
    elapsed = time.perf_counter() - start
    return result, elapsed

def gpu_memory_gb():
    """Modified GPU memory reporting function for Apple Silicon Mac"""
    try:
        import torch
        if torch.cuda.is_available():
            return {
                "device": "cuda",
                "allocated_GB": round(torch.cuda.memory_allocated() / 1024**3, 3),
                "reserved_GB": round(torch.cuda.memory_reserved() / 1024**3, 3),
                "max_allocated_GB": round(torch.cuda.max_memory_allocated() / 1024**3, 3),
            }
        elif hasattr(torch.backends, "mps") and torch.backends.mps.is_available():
            return {
                "device": "mps",
                "allocated_GB": round(torch.mps.current_allocated_memory() / 1024**3, 3),
                "driver_allocated_GB": round(torch.mps.driver_allocated_memory() / 1024**3, 3),
            }
        return None
    except Exception:
        return None
print("GPU memory snapshot:", gpu_memory_gb())
```
